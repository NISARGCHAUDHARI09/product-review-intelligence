```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd()

sys.path.append(
    str(PROJECT_ROOT / "src")
)

from inference import SentimentPredictor


MODEL_DIR = (
    PROJECT_ROOT
    / "models"
    / "final_distilbert_model"
)


predictor = SentimentPredictor(
    model_path=MODEL_DIR
)
```

```python
tests = [
    "This product is absolutely fantastic. I love it.",
    "This product is terrible. I hate it.",
    "The quality is okay, nothing special.",
    "Amazing product, excellent quality and very useful.",
    "Worst purchase ever. Completely disappointing."
]

for text in tests:

    result = predictor.predict(text)

    print(
        f"{result['label']:8s} "
        f"{result['confidence']:.4f} | "
        f"{text}"
    )
```

```python
results = predictor.predict_batch(tests)

for text, result in zip(tests, results):

    print(
        result["label"],
        f"{result['confidence']:.4f}",
        "|",
        text
    )
```
